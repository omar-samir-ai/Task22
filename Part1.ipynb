## Part 1: Research 

#### 1. What is data cleaning, and why is it important?
Data cleaning is the process of detecting and correcting (or removing) corrupt, inaccurate, duplicate, or incomplete records from a dataset. It is important because clean data ensures high accuracy, prevents misleading analytical results, and improves the overall performance of machine learning models.

---

#### 2. How do isnull() and isna() identify missing values?
Both `isnull()` and `isna()` function identically in Pandas to detect missing or null values (such as `NaN` or `None`). They return a Boolean DataFrame of the same shape, where `True` indicates a missing value and `False` indicates a valid value.

---

#### 3. What is the difference between dropna() and fillna()?
- `dropna()`: Removes/drops rows or columns that contain missing values based on specified conditions.
- `fillna()`: Replaces missing values with a specified value (like 0, a mean, or a specific placeholder) instead of deleting the data.

---

#### 4. How does drop_duplicates() work?
`drop_duplicates()` scans the DataFrame for identical rows and removes the duplicates, keeping either the first or the last occurrence based on the `keep` parameter (default is `keep='first'`).

---

#### 5. How can Pandas convert columns to appropriate data types using astype() and to_datetime()?
- `astype()`: Converts a column's data type to a specified standard type (e.g., converting a numeric column stored as string into integers or floats).
- `to_datetime()`: Specifically parses date and time strings into proper Pandas timestamp objects (`datetime64`), enabling correct date-based operations and filtering.
